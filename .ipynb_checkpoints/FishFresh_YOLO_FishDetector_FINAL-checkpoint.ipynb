# 🐟 FishFresh — YOLO Fish Detector → TFLite (FINAL)

This notebook loads a **pretrained YOLOv8 (OpenImages‑V7)** model, filters outputs to **Fish** only, predicts on your **tray photos** under `assets/samples/`, and exports **both** FP16 and FP32 **TFLite** files for your Flutter app. It also includes a **test/evaluation from assets** with optional labels and writes a **per‑image CSV** summary.

**Flow:** Detect (Fish only) → crop (in app) → classify with your MobileNet (species + freshness).

## 1) Paths & folders
Put a few tray images (jpg/png) in **`assets/samples/`** to test.

````python
from pathlib import Path

ROOT = Path.cwd()

# All assets inside model/
MODEL_DIR = ROOT / 'model'

ASSETS  = MODEL_DIR / 'assets'
ASSETS.mkdir(parents=True, exist_ok=True)

SAMPLES = ASSETS / 'samples'
SAMPLES.mkdir(parents=True, exist_ok=True)

# Output folder also inside model/
OUT = MODEL_DIR / 'runs_fishdet'
OUT.mkdir(parents=True, exist_ok=True)

print("ROOT    =", ROOT)
print("MODEL   =", MODEL_DIR)
print("ASSETS  =", ASSETS)
print("SAMPLES =", SAMPLES, "(put tray photos here)")
print("OUT     =", OUT)
````

## 2) Load YOLOv8 (OpenImages‑V7) and resolve `fish_id`
- `yolov8n-oiv7.pt` = fastest/smallest
- `yolov8s-oiv7.pt` = a bit more accurate (slower)

````python
from ultralytics import YOLO

MODEL_NAME = 'yolov8s-oiv7.pt'  # or 'yolov8s-oiv7.pt'
model = YOLO(MODEL_NAME)        # auto-downloads on first use
print('[✓] Loaded YOLO model:', MODEL_NAME)

# Resolve the class id for 'Fish' (case-insensitive)
fish_id = next((int(k) for k, v in model.names.items() if str(v).lower() == 'fish'), None)
assert fish_id is not None, "Fish class not found in this checkpoint."
print('Fish class id =', fish_id)
````

## 3) Predict on assets/samples (Fish only)
Tune `IMGSZ`, `CONF`, `MAXDET` as needed. Annotated results saved under `runs_fishdet/pred/`.

````python
IMGSZ, CONF, MAXDET = 768, 0.10, 400

img_list = [*SAMPLES.glob('*.jpg'), *SAMPLES.glob('*.jpeg'), *SAMPLES.glob('*.png')]
print('Found', len(img_list), 'images in', SAMPLES)

if img_list:
    results = model.predict(
        source=str(SAMPLES),
        imgsz=IMGSZ,
        conf=CONF,            # low; evaluation cell will pick best threshold
        iou=0.60,             # slightly tighter NMS for tray piles
        max_det=MAXDET,
        classes=[fish_id],    # fish-only
        save=False,            # set False when not inspecting images
        save_txt=False,        # set False unless you need labels
        project=str(OUT),
        name='pred',
        exist_ok=True,
        verbose=False,
        workers=0,            # Windows-safe
        # device=0            # uncomment if you have CUDA
    )
    print('[✓] Done. See annotated results in:', OUT/'pred')
else:
    print('[i] No images found in', SAMPLES)
````

## 4) Export **both** TFLite files + config
Writes to `runs_fishdet/export/`:
- `fishdet_fp16.tflite`
- `fishdet_fp32.tflite`
- `detector_config.json`

````python
from pathlib import Path
import shutil
from datetime import datetime
import json

# === SETTINGS ===
IMGSZ = int(globals().get("IMGSZ", 768))
MAXDET = int(globals().get("MAXDET", 50))
DISPLAY_CONF = float(globals().get("DISPLAY_CONF", 0.35))

# === TIMESTAMP ===
timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")     # For filenames
human_time = datetime.now().strftime("%A, %B %d, %Y at %I:%M:%S %p")  # For printing

print("========================================")
print(" Exporting ONNX + detector_config.json")
print(" Image Size :", IMGSZ)
print(" Export Time:", human_time)
print("========================================\n")

# === EXPORT ONNX ===
res = model.export(
    format="onnx",
    imgsz=IMGSZ,
    opset=17,
    simplify=True,
    dynamic=False
)

# === FIND ONNX OUTPUT ===
onnx_file = None
for p in res if isinstance(res, (list, tuple)) else [res]:
    p = Path(p)
    if p.suffix == ".onnx" and p.exists():
        onnx_file = p
        break

assert onnx_file is not None, "ONNX export succeeded but ONNX file not found."

# === SAVE TO export/ WITH TIMESTAMP ===
EXPORT_DIR = Path("model/runs_fishdet/export")
EXPORT_DIR.mkdir(parents=True, exist_ok=True)

onnx_out = EXPORT_DIR / f"fishdet_yolov8s_oiv7_{IMGSZ}_{timestamp}.onnx"
shutil.copy2(onnx_file, onnx_out)

print("✓ ONNX exported to:", onnx_out)

# === FLUTTER DETECTOR CONFIG JSON ===
# NOTE: tflite_asset is the path you will use in pubspec.yaml
# e.g.  assets/model/fishdet_fp16.tflite
detector_config = {
    "detector_name": f"fishdet_yolov8s_oiv7_{IMGSZ}",
    "input_size": IMGSZ,
    "conf_threshold": DISPLAY_CONF,   # from your eval code if set
    "iou_threshold": 0.60,
    "max_detections": MAXDET,
    "tflite_asset": "assets/model/fishdet_fp16.tflite",  # <--- Flutter asset path
    "label_map": {
        "0": "fish"   # YOLO detector = just "fish" (you crop, Mobilenet does species+freshness)
    },
    "export_info": {
        "onnx_file": onnx_out.name,
        "export_time": human_time
    }
}

json_out = EXPORT_DIR / "fishdet_config.json"
with open(json_out, "w") as f:
    json.dump(detector_config, f, indent=2)

print("✓ detector_config.json exported to:", json_out)
print("✓ Completed at:", human_time)
````

## 6) Evaluate directly from **assets/**
Two modes:
- **With labels** at `assets/labels/*.txt` (YOLO format; class `0 = fish`) → reports **mAP@0.5**, **Precision/Recall/F1** at your chosen `CONF_EVAL`, and writes a **per‑image CSV**.
- **Without labels** → reports **Detection Rate**, **avg detections/image**, **avg confidence** and writes a per‑image CSV with detections only.

> You can adjust `CONF_EVAL` below (default **0.35**).

````python
# === Evaluate on assets/ — mAP@0.5 + best-F1 & summary ===
from pathlib import Path
from PIL import Image
import numpy as np, csv

# paths
IMG_DIR = SAMPLES                 # assets/samples
LBL_DIR = ASSETS/'labels'         # optional labels in YOLO format (class 0 = fish)
CSV_PATH = EXPORT_DIR/'assets_eval_summary.csv'

# knobs
DISPLAY_CONF = 0.35               # what you want to display to your adviser
IOU_MATCH    = 0.50               # IoU threshold for AP and TP/FP
IMGSZ_EVAL   = IMGSZ              # 768 now
MAXDET_EVAL  = MAXDET

img_paths = sorted([*IMG_DIR.glob('*.jpg'), *IMG_DIR.glob('*.jpeg'), *IMG_DIR.glob('*.png')])
assert len(img_paths) > 0, f'No images in {IMG_DIR}'

has_labels = LBL_DIR.exists() and any(LBL_DIR.glob('*.txt'))

def yolo_txt_to_boxes(lbl_path, W, H):
    if not lbl_path.exists():
        return []
    boxes = []
    for line in lbl_path.read_text().strip().splitlines():
        p = line.split()
        if len(p) < 5:
            continue
        # ignore class id, just use bbox
        _, cx, cy, w, h = p[:5]
        cx, cy, w, h = map(float, (cx, cy, w, h))
        x1 = (cx - w/2) * W; y1 = (cy - h/2) * H
        x2 = (cx + w/2) * W; y2 = (cy + h/2) * H
        boxes.append([x1, y1, x2, y2])
    return boxes

def iou_xyxy(a, b):
    ax1, ay1, ax2, ay2 = a; bx1, by1, bx2, by2 = b
    ix1, iy1 = max(ax1,bx1), max(ay1,by1)
    ix2, iy2 = min(ax2,bx2), min(ay2,by2)
    iw, ih = max(0.0, ix2-ix1), max(0.0, iy2-iy1)
    inter = iw * ih
    ua = (ax2-ax1)*(ay2-ay1) + (bx2-bx1)*(by2-by1) - inter + 1e-6
    return inter / ua

# 1) Run detector with low conf; we will sweep thresholds later
all_preds = []
for p in img_paths:
    r = model.predict(
        source=str(p),
        imgsz=IMGSZ_EVAL, conf=0.10, iou=0.6, max_det=MAXDET_EVAL,
        # classes=[fish_id],  # <-- REMOVE THIS
        verbose=False, save=False, save_txt=False, save_conf=False
    )[0]
    if r.boxes is None or len(r.boxes) == 0:
        preds = np.zeros((0,5), dtype=np.float32)
    else:
        xyxy   = r.boxes.xyxy.cpu().numpy()
        scores = r.boxes.conf.cpu().numpy().reshape(-1,1)
        preds  = np.hstack([xyxy, scores]).astype(np.float32)
    all_preds.append(preds)

if not has_labels:
    # 2) Stats without labels (no accuracy possible)
    def quick(th):
        det_images = 0; sum_dets = 0; sum_conf = 0.0
        for preds in all_preds:
            sel = preds[preds[:,4] >= th]
            n = sel.shape[0]
            if n > 0:
                det_images += 1
                sum_dets += n
                sum_conf += float(sel[:,4].mean())
        det_rate = det_images / len(img_paths)
        avg_dets = sum_dets / len(img_paths)
        avg_conf = (sum_conf / max(1, det_images)) if det_images else 0.0
        return det_rate, avg_dets, avg_conf

    dr, ad, ac = quick(DISPLAY_CONF)
    print('=== Quick check (no labels found) ===')
    print(f'Images: {len(img_paths)}')
    print(f'Detection Rate (≥1 fish): {dr*100:.2f}% @ conf ≥ {DISPLAY_CONF:.2f}')
    print(f'Avg detections per image: {ad:.2f}')
    print(f'Avg confidence (detected imgs): {ac*100:.1f}%')

    print('\nThreshold sweep (conf → DetRate  AvgDets  AvgConf)')
    for t in [0.10,0.20,0.25,0.30,0.35,0.40]:
        dr, ad, ac = quick(t)
        print(f"{t:>5.2f} → {dr*100:6.2f}%   {ad:6.2f}   {ac*100:6.1f}%")

    rows = []
    for p, preds in zip(img_paths, all_preds):
        sel = preds[preds[:,4] >= DISPLAY_CONF]
        rows.append([p.name, sel.shape[0], float(sel[:,4].mean()) if len(sel) else 0.0])
    with open(CSV_PATH, 'w', newline='') as f:
        w = csv.writer(f); w.writerow(['image','num_detections_at_conf','avg_confidence_at_conf'])
        w.writerows(rows)
    print('\n[✓] Wrote CSV:', CSV_PATH)

else:
    # 3) Full metrics with labels
    all_gts = []
    for p in img_paths:
        with Image.open(p) as im:
            W, H = im.size
        all_gts.append(yolo_txt_to_boxes(LBL_DIR/(p.stem+'.txt'), W, H))

    # PR curve → AP50
    def pr_at(th):
        TP=FP=FN=0
        for gts, preds in zip(all_gts, all_preds):
            sel = preds[preds[:,4] >= th]
            used = set()
            for g in gts:
                best=-1; best_i=-1
                for i,b in enumerate(sel[:, :4]):
                    if i in used: continue
                    iou = iou_xyxy(g, b)
                    if iou > best:
                        best, best_i = iou, i
                if best >= IOU_MATCH and best_i >= 0:
                    TP += 1; used.add(best_i)
                else:
                    FN += 1
            FP += len(sel) - len(used)
        prec = TP/(TP+FP+1e-6); rec = TP/(TP+FN+1e-6)
        return prec, rec

    ths = np.linspace(0.0, 1.0, 101)
    PR = np.array([pr_at(t) for t in ths])
    prec, rec = PR[:,0], PR[:,1]

    # AP computation with monotonic precision
    order = np.argsort(rec)
    rec_sorted  = rec[order]
    prec_sorted = prec[order].copy()
    for i in range(len(prec_sorted)-2, -1, -1):
        prec_sorted[i] = max(prec_sorted[i], prec_sorted[i+1])
    ap50 = float(np.trapz(prec_sorted, rec_sorted))

    # Find best F1 threshold
    F1 = (2*prec*rec)/(prec+rec+1e-6)
    best_i   = int(np.nanargmax(F1))
    BEST_CONF = float(ths[best_i])
    best_P, best_R, best_F1 = float(prec[best_i]), float(rec[best_i]), float(F1[best_i])

    # Also compute metrics at DISPLAY_CONF for reporting
    def metrics_at(th):
        TP=FP=FN=0
        rows=[]
        for p, gts, preds in zip(img_paths, all_gts, all_preds):
            sel = preds[preds[:,4] >= th]
            used=set()
            for g in gts:
                best=-1; best_i=-1
                for i,b in enumerate(sel[:, :4]):
                    if i in used: continue
                    iou = iou_xyxy(g,b)
                    if iou > best:
                        best, best_i = iou, i
                if best >= IOU_MATCH and best_i >= 0:
                    TP += 1; used.add(best_i)
                else:
                    FN += 1
            FP += len(sel) - len(used)
            rows.append([p.name, len(gts), len(sel), len(used), len(sel)-len(used)])
        P = TP/(TP+FP+1e-6); R = TP/(TP+FN+1e-6); F1 = 2*P*R/(P+R+1e-6)
        return P, R, F1, rows

    P_disp, R_disp, F1_disp, rows = metrics_at(DISPLAY_CONF)

    print('=== Labeled evaluation (YOLO TXT in assets/labels) ===')
    print(f'Images: {len(img_paths)}')
    print(f'mAP@0.5: {ap50*100:.2f}%')
    print(f'Best F1 at conf={BEST_CONF:.2f} → P={best_P*100:.2f}%  R={best_R*100:.2f}%  F1={best_F1*100:.2f}%')
    print(f'At conf={DISPLAY_CONF:.2f} → P={P_disp*100:.2f}%  R={R_disp*100:.2f}%  F1={F1_disp*100:.2f}%')

    with open(CSV_PATH, 'w', newline='') as f:
        w = csv.writer(f)
        w.writerow(['image','gt_fish','pred_fish_at_conf','TP_at_conf','FP_at_conf'])
        w.writerows(rows)
    print('[✓] Wrote CSV:', CSV_PATH)

# Tip: after this, write BEST_CONF into detector_config.json for app inference if desired
````
