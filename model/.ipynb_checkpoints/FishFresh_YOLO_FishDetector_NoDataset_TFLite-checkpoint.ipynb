## 3) Point to your **fish detector** weights (`.pt`)
**Option A (recommended):** Paste a **direct download URL** to a YOLO `.pt` file (from Roboflow Universe or other source).  
**Option B:** Leave the URL empty and put your `.pt` **locally** at `assets/fish_detector.pt`.

> ⚠️ These should be **detection** weights (YOLOv8/YOLOv11) trained for a **single class** `fish`. If you only have YOLOv5 weights, you can convert with Ultralytics, but using v8/v11 `.pt` is simpler.


```python
# === Paths & folders (run this once) ===
from pathlib import Path

ROOT    = Path.cwd()
ASSETS  = ROOT / 'assets';  ASSETS.mkdir(exist_ok=True)
SAMPLES = ASSETS / 'samples'; SAMPLES.mkdir(parents=True, exist_ok=True)
OUT     = ROOT / 'runs_fishdet'; OUT.mkdir(exist_ok=True)

print("ROOT   =", ROOT)
print("ASSETS =", ASSETS)
print("SAMPLES=", SAMPLES, "(put your tray photos here)")
print("OUT    =", OUT)
```

```python
# Opens the samples folder in File Explorer (Windows only)
import os, sys
if sys.platform.startswith("win"):
    os.startfile(str(SAMPLES))
else:
    print("Open this folder manually:", SAMPLES)
```

```python
# Lists a few images found in assets/samples
img_list = [*SAMPLES.glob("*.jpg"), *SAMPLES.glob("*.jpeg"), *SAMPLES.glob("*.png")]
print("Found", len(img_list), "images")
for p in img_list[:10]:
    print(" -", p.name)
```

```python
# === Use an Ultralytics built-in model (auto-download, no URL needed) ===
# 'yolov8n-oiv7.pt' = smallest/fast; 'yolov8s-oiv7.pt' = a bit stronger.
from ultralytics import YOLO

MODEL_NAME = 'yolov8n-oiv7.pt'   # or 'yolov8s-oiv7.pt'
model = YOLO(MODEL_NAME)         # auto-downloads on first use
print('[✓] Loaded YOLO model:', MODEL_NAME)

# Find the class id for "Fish" so we can filter to fish-only
fish_id = None
for k, v in model.names.items():
    if str(v).lower() == 'fish':
        fish_id = int(k); break

assert fish_id is not None, "Fish class not found in this checkpoint."
print('Fish class id =', fish_id)
```

```python
# --- Predict on tray photos (fish-only) ---
IMGSZ, CONF, MAXDET = 640, 0.30, 300

if any(SAMPLES.glob('*.*')):
    results = model.predict(
        source=str(SAMPLES),
        imgsz=IMGSZ,
        conf=CONF,
        max_det=MAXDET,
        classes=[fish_id],      # <- only return FISH boxes
        save=True,
        save_txt=True,
        project=str(OUT),
        name='pred',
        exist_ok=True,
        verbose=False,
    )
    print('[✓] Done. See annotated results in:', OUT/'pred')
else:
    print('[i] No images found in', SAMPLES)
```

## 6) Export to **TFLite** (choose FP32 or FP16)
- **FP32**: bigger model, most compatible.  
- **FP16**: smaller model, faster on some devices with float16 support.  
- **INT8**: requires calibration images & a representative dataset (not covered here since we’re not building a dataset).


```python
# Choose your export precision
EXPORT_FP16 = True   # set False for FP32
IMGSZ = 640          # must match your Flutter pre/post pipeline assumptions

print('[*] Exporting to TFLite... (this can take a few minutes)')
export_path = model.export(
    format='tflite',
    imgsz=IMGSZ,
    half=EXPORT_FP16,   # True -> FP16, False -> FP32
    int8=False,         # No calibration in this notebook
)
print('[✓] Exported file:', export_path)
```

```python
# === Export BOTH FP16 and FP32 TFLite files ===
from pathlib import Path
import shutil

IMGSZ = 640
EXPORT_DIR = OUT / "export"
EXPORT_DIR.mkdir(parents=True, exist_ok=True)

print("[*] Exporting FP16 (half=True)...")
res16 = model.export(format="tflite", imgsz=IMGSZ, half=True, int8=False)
tfl16 = next((Path(p) for p in (res16 if isinstance(res16, (list, tuple)) else [res16]) if str(p).endswith(".tflite")), None)
assert tfl16 and tfl16.exists(), "FP16 TFLite not found."
fp16_out = EXPORT_DIR / "fishdet_fp16.tflite"
shutil.copy2(tfl16, fp16_out)
print("[✓] FP16 saved to:", fp16_out)

print("[*] Exporting FP32 (half=False)...")
res32 = model.export(format="tflite", imgsz=IMGSZ, half=False, int8=False)
tfl32 = next((Path(p) for p in (res32 if isinstance(res32, (list, tuple)) else [res32]) if str(p).endswith(".tflite")), None)
assert tfl32 and tfl32.exists(), "FP32 TFLite not found."
fp32_out = EXPORT_DIR / "fishdet_fp32.tflite"
shutil.copy2(tfl32, fp32_out)
print("[✓] FP32 saved to:", fp32_out)

print("\\nAll exports:")
print(" -", fp16_out)
print(" -", fp32_out)
```

## 7) (Optional) Quick local test of the exported `.tflite`
We’ll load the `.tflite` with TensorFlow Lite Interpreter and run a dry pass on one image to confirm it executes.


```python
import tensorflow as tf
from PIL import Image
import numpy as np

# Find the exported file
exported = None
if isinstance(export_path, (list, tuple)):
    for p in export_path:
        if str(p).endswith('.tflite'): exported = Path(p)
else:
    p = Path(export_path)
    exported = p if p.suffix == '.tflite' else None

assert exported and exported.exists(), 'TFLite file not found after export.'
print('[i] Using TFLite:', exported)

# Pick any sample image if available
img_files = [*SAMPLES.glob('*.jpg'), *SAMPLES.glob('*.jpeg'), *SAMPLES.glob('*.png')]
if not img_files:
    print('[i] No sample image to test. Skipping interpreter run.')
else:
    test_img = img_files[0]
    print('[*] Testing with:', test_img.name)

    # Preprocess (match imgsz and normalization expected by exported model)
    im = Image.open(test_img).convert('RGB').resize((IMGSZ, IMGSZ))
    arr = np.array(im, dtype=np.float32)
    arr = arr / 255.0  # YOLO exports expect 0-1 range
    arr = np.expand_dims(arr, axis=0)

    # Run interpreter
    interpreter = tf.lite.Interpreter(model_path=str(exported))
    interpreter.allocate_tensors()
    input_details = interpreter.get_input_details()
    output_details = interpreter.get_output_details()

    # Some exports expect NHWC float32/16
    interpreter.set_tensor(input_details[0]['index'], arr)
    interpreter.invoke()

    # Show output tensor shapes
    print('[✓] TFLite invoked successfully.')
    for od in output_details:
        out = interpreter.get_tensor(od['index'])
        print(' - output name:', od.get('name'), 'shape:', out.shape, 'dtype:', out.dtype)
```

## 8) Flutter integration notes (detector → crop → MobileNet classifier)
- **Detector input size**: use the same `imgsz` (e.g., 640).  
- **Preprocess**: resize, keep aspect ratio (letterbox) or center-crop, **normalize 0–1**.  
- **Outputs**: YOLO TFLite exports typically return **boxes + scores + class ids**; you’ll need NMS (some exports include built-in NMS).  
- **Pipeline**:
  1. Run detector TFLite on the tray photo.
  2. For each box with class=`fish` and score ≥ threshold, crop that ROI from the **original** image.
  3. Send each crop to your **MobileNet classifier** to get *(species → freshness)*.

**Tip:** Start with FP16 export; if your device has issues, switch to FP32.

**Where is the file?** After export, look under a new folder like `\n`- `weights/best.tflite` (Ultralytics writes paths in the cell output).

---
### Troubleshooting
- **Export fails with missing packages** → re-run the install cell; ensure `ultralytics`, `onnx`, and `tensorflow-cpu` are installed.
- **TFLite not working on device** → try FP32; confirm your Flutter pre/post-processing matches the notebook.
- **Detector misses crowded/overlapping fish** → later, fine-tune with ~200–500 corrected boxes from your market photos.
