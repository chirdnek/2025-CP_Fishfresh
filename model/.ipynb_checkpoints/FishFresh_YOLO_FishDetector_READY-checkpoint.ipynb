# FishFresh — YOLO Fish Detector (No-Dataset) → ONNX → TFLite → Evaluate

This notebook loads a **pretrained YOLOv8 (OIv7)** fish-capable model, filters to the fish class,
exports **ONNX**, runs your **external .bat converter** to produce **TFLite (FP32 & FP16)**, optionally checks that the TFLite files load, and provides a **clean evaluation** that prints
**mAP@0.5 / Precision / Recall / F1** when labels exist (or a detection-rate summary when labels are absent).

```python
# === 0) Paths & runtime ===
from pathlib import Path
import os, json, shutil, sys

ROOT = Path.cwd()               # notebook folder
ASSETS = ROOT / 'assets'
SAMPLES = ASSETS / 'samples'    # images for quick evaluation
OUT = ROOT / 'runs_fishdet'
EXPORT_DIR = OUT / 'export'
OUT.mkdir(parents=True, exist_ok=True); EXPORT_DIR.mkdir(parents=True, exist_ok=True)

# Safer temp on Windows for pip / exports
Path('C:/piptmp').mkdir(parents=True, exist_ok=True)
os.environ['TMP']  = r'C:/piptmp'
os.environ['TEMP'] = r'C:/piptmp'

print('ROOT =', ROOT)
print('SAMPLES exists:', SAMPLES.exists())
print('EXPORT_DIR =', EXPORT_DIR)
```

```python
# === 1) Install minimal deps (one-time) ===
%pip -q install --upgrade --prefer-binary ultralytics onnx onnxruntime tensorflow-cpu --root-user-action=ignore
```

```python
# === 2) Load built-in model & find fish class ===
from ultralytics import YOLO

MODEL_NAME = 'yolov8s-oiv7.pt'   # 'yolov8n-oiv7.pt' for even lighter
IMGSZ = 640                      # try 768/960 if tiny fish are missed
MAXDET = 300

model = YOLO(MODEL_NAME)
print('[✓] Loaded YOLO model:', MODEL_NAME)

fish_id = None
for k, v in model.model.names.items():
    if str(v).lower() == 'fish':
        fish_id = int(k); break
assert fish_id is not None, 'Fish class not found!'
print('Fish class id =', fish_id)
```

```python
# === 3) Optional: quick dry-run on one sample ===
imgs = sorted([*SAMPLES.glob('*.jpg'), *SAMPLES.glob('*.jpeg'), *SAMPLES.glob('*.png')])
if imgs:
    r = model.predict(
        str(imgs[0]), imgsz=IMGSZ, conf=0.25, max_det=MAXDET, classes=[fish_id],
        verbose=False, save=False, save_txt=False, save_conf=False
    )[0]
    print('[i] Boxes:', (0 if r.boxes is None else len(r.boxes)))
else:
    print('[i] No sample images in', SAMPLES)
```

```python
# === 4) Export ONNX (pre-NMS) + write tiny config ===
from pathlib import Path
import json, shutil

onnx_out = EXPORT_DIR / 'fishdet_yolov8s_oiv7.onnx'
print('[*] Exporting ONNX to:', onnx_out)

res = model.export(format='onnx', imgsz=IMGSZ, half=False, dynamic=False, simplify=True)

# Pick the ONNX from whatever Ultralytics returned
out_path = None
for p in (res if isinstance(res, (list,tuple)) else [res]):
    p = Path(str(p))
    if p.suffix.lower() == '.onnx' and p.exists():
        out_path = p; break
assert out_path and out_path.exists(), 'ONNX export failed.'
if out_path.resolve() != onnx_out.resolve():
    shutil.copy2(out_path, onnx_out)

# Tiny config for your app
(EXPORT_DIR/'detector_config.json').write_text(
    json.dumps({'imgsz': int(IMGSZ), 'conf': 0.30, 'fish_id': int(fish_id)}, indent=2)
)
print('[✓] Wrote:', EXPORT_DIR/'detector_config.json')
print('All exports:\n -', onnx_out)
```

```python
# === 4b) Run your external ONNX→TFLite converter and copy expected names ===
import subprocess, shutil
from pathlib import Path

BAT_PATH = Path(r'C:\\aziztebbeng\\2025-CP_Fishfresh\\model\\YOLO_Converter\\convert_yolov8_onnx_to_tflite.bat')
CONV_DIR = BAT_PATH.parent
OUT_DIR  = CONV_DIR / 'out_tflite'

assert BAT_PATH.exists(), f'Converter .bat not found: {BAT_PATH}'
assert (EXPORT_DIR/'fishdet_yolov8s_oiv7.onnx').exists(), 'ONNX not found — run export cell.'

cmd = f'"{BAT_PATH}" "{EXPORT_DIR/"fishdet_yolov8s_oiv7.onnx"}" "{OUT_DIR}"'
print('[*] Running converter:\n ', cmd, '\n')
proc = subprocess.run(cmd, shell=True, cwd=str(CONV_DIR))
if proc.returncode != 0:
    raise RuntimeError(f'Converter failed with exit code {proc.returncode}')

# Your converter writes: yolov8_fp32.tflite / yolov8_fp16.tflite
tfl_fp32 = OUT_DIR / 'yolov8_fp32.tflite'
tfl_fp16 = OUT_DIR / 'yolov8_fp16.tflite'
assert tfl_fp32.exists() and tfl_fp16.exists(), 'Converter outputs not found.'

# Copy + rename to what the notebook expects next
shutil.copy2(tfl_fp32, EXPORT_DIR/'fishdet_fp32.tflite')
shutil.copy2(tfl_fp16, EXPORT_DIR/'fishdet_fp16.tflite')

print('[✓] TFLite ready:')
print(' -', EXPORT_DIR/'fishdet_fp16.tflite')
print(' -', EXPORT_DIR/'fishdet_fp32.tflite')
```

```python
# === 5) (Optional) TFLite load test ===
import tensorflow as tf
for p in [EXPORT_DIR/'fishdet_fp16.tflite', EXPORT_DIR/'fishdet_fp32.tflite']:
    print('\n[i] Testing:', p)
    interpreter = tf.lite.Interpreter(model_path=str(p))
    interpreter.allocate_tensors()
    outs = interpreter.get_output_details()
    print('[✓] Interpreter OK. Outputs:')
    for od in outs:
        print(' -', od.get('name'), od['shape'], od['dtype'])
```

```python
# === 6) Evaluate directly from assets/ — clean metrics & table ===
from pathlib import Path
from PIL import Image
import numpy as np, csv

IMGSZ_EVAL   = int(globals().get('IMGSZ', 640))
CONF_EVAL    = float(globals().get('CONF_EVAL', 0.35))  # operating threshold
MAXDET_EVAL  = int(globals().get('MAXDET', 300))

IMG_DIR = SAMPLES                 # assets/samples
LBL_DIR = ASSETS/'labels'         # optional labels folder
CSV_PATH = EXPORT_DIR/'assets_eval_summary.csv'

img_paths = sorted([*IMG_DIR.glob('*.jpg'), *IMG_DIR.glob('*.jpeg'), *IMG_DIR.glob('*.png')])
assert len(img_paths) > 0, f'No images in {IMG_DIR}'

has_labels = LBL_DIR.exists() and any(LBL_DIR.glob('*.txt'))

def _yolo_txt_to_boxes(lbl_path, W, H):
    if not lbl_path.exists(): return []
    boxes = []
    for line in lbl_path.read_text().strip().splitlines():
        p = line.split()
        if len(p) != 5: 
            continue
        cls, cx, cy, w, h = p
        if int(float(cls)) != 0: 
            continue  # single-class fish
        cx, cy, w, h = map(float, (cx, cy, w, h))
        x1 = (cx - w/2) * W; y1 = (cy - h/2) * H
        x2 = (cx + w/2) * W; y2 = (cy + h/2) * H
        boxes.append([x1,y1,x2,y2])
    return boxes

def _iou_xyxy(a, b):
    ax1, ay1, ax2, ay2 = a; bx1, by1, bx2, by2 = b
    ix1, iy1 = max(ax1,bx1), max(ay1,by1)
    ix2, iy2 = min(ax2,bx2), min(ay2,by2)
    iw, ih = max(0.0, ix2-ix1), max(0.0, iy2-iy1)
    inter = iw * ih
    ua = (ax2-ax1)*(ay2-ay1) + (bx2-bx1)*(by2-by1) - inter + 1e-6
    return inter / ua

# 1) Run detector on all images (quiet; no saving)
all_preds = []
for p in img_paths:
    r = model.predict(
        source=str(p), imgsz=IMGSZ_EVAL, conf=0.10,
        max_det=MAXDET_EVAL, classes=[fish_id],
        verbose=False, save=False, save_txt=False, save_conf=False
    )[0]
    if r.boxes is None or len(r.boxes) == 0:
        preds = np.zeros((0,5), dtype=np.float32)
    else:
        xyxy   = r.boxes.xyxy.cpu().numpy()
        scores = r.boxes.conf.cpu().numpy().reshape(-1,1)
        preds  = np.hstack([xyxy, scores]).astype(np.float32)
    all_preds.append(preds)

# 2) No labels: quick stats + sweep
if not has_labels:
    def quick_stats(th):
        det_images = 0; sum_dets = 0; sum_conf = 0.0
        for preds in all_preds:
            sel = preds[preds[:,4] >= th]
            n = sel.shape[0]
            if n > 0:
                det_images += 1
                sum_dets += n
                sum_conf += float(sel[:,4].mean())
        det_rate = det_images / len(img_paths)
        avg_dets = sum_dets / len(img_paths)
        avg_conf = (sum_conf / max(1, det_images)) if det_images else 0.0
        return det_rate, avg_dets, avg_conf

    dr, ad, ac = quick_stats(CONF_EVAL)
    print('=== Assets quick check (no labels found) ===')
    print(f'Images: {len(img_paths)}')
    print(f'Detection Rate (≥1 fish): {dr*100:.2f}%  at conf ≥ {CONF_EVAL:.2f}')
    print(f'Avg detections per image: {ad:.2f}')
    print(f'Avg confidence (detected imgs): {ac*100:.1f}%')

    print('\nThreshold sweep (conf → DetRate  AvgDets  AvgConf)')
    for t in [0.10,0.20,0.25,0.30,0.35,0.40]:
        dr, ad, ac = quick_stats(t)
        print(f"{t:>5.2f} → {dr*100:6.2f}%   {ad:6.2f}   {ac*100:6.1f}%")

    # per-image CSV
    rows = []
    for p, preds in zip(img_paths, all_preds):
        sel = preds[preds[:,4] >= CONF_EVAL]
        rows.append([p.name, sel.shape[0], float(sel[:,4].mean()) if len(sel) else 0.0])
    with open(CSV_PATH, 'w', newline='') as f:
        w = csv.writer(f); w.writerow(['image','num_detections_at_conf','avg_confidence_at_conf'])
        w.writerows(rows)
    print('\n[✓] Wrote CSV:', CSV_PATH)

# 3) With labels: mAP@0.5 + P/R/F1 at CONF_EVAL
else:
    IOU = 0.50
    all_gts = []
    for p in img_paths:
        with Image.open(p) as im:
            W,H = im.size
        all_gts.append(_yolo_txt_to_boxes(LBL_DIR/(p.stem+'.txt'), W, H))

    def pr_at(th):
        TP=FP=FN=0
        for gts, preds in zip(all_gts, all_preds):
            sel = preds[preds[:,4] >= th]
            used = set()
            for g in gts:
                best=-1; best_i=-1
                for i,b in enumerate(sel[:, :4]):
                    if i in used: continue
                    iou = _iou_xyxy(g, b)
                    if iou > best:
                        best, best_i = iou, i
                if best >= IOU and best_i >= 0:
                    TP += 1; used.add(best_i)
                else:
                    FN += 1
            FP += len(sel) - len(used)
        prec = TP/(TP+FP+1e-6); rec = TP/(TP+FN+1e-6)
        return prec, rec

    ths = np.linspace(0.0, 1.0, 101)
    pr  = np.array([pr_at(t) for t in ths])
    prec, rec = pr[:,0], pr[:,1]

    order = np.argsort(rec)
    rec_sorted  = rec[order]
    prec_sorted = prec[order].copy()
    for i in range(len(prec_sorted)-2, -1, -1):
        prec_sorted[i] = max(prec_sorted[i], prec_sorted[i+1])
    ap50 = float(np.trapz(prec_sorted, rec_sorted))

    TP=FP=FN=0; rows=[]
    for p, gts, preds in zip(img_paths, all_gts, all_preds):
        sel = preds[preds[:,4] >= CONF_EVAL]
        used=set()
        for g in gts:
            best=-1; best_i=-1
            for i,b in enumerate(sel[:, :4]):
                if i in used: continue
                iou = _iou_xyxy(g,b)
                if iou > best:
                    best, best_i = iou, i
            if best >= IOU and best_i >= 0:
                TP += 1; used.add(best_i)
            else:
                FN += 1
        FP += len(sel) - len(used)
        rows.append([p.name, len(gts), len(sel), len(used), len(sel)-len(used)])

    P = TP/(TP+FP+1e-6); R = TP/(TP+FN+1e-6); F1 = 2*P*R/(P+R+1e-6)

    print('=== Assets evaluation (labels found) ===')
    print(f'Images: {len(img_paths)}')
    print(f'mAP@0.5: {ap50*100:.2f}%')
    print(f'@conf={CONF_EVAL:.2f} → Precision={P*100:.2f}%  Recall={R*100:.2f}%  F1={F1*100:.2f}%')

    with open(CSV_PATH, 'w', newline='') as f:
        w = csv.writer(f)
        w.writerow(['image','gt_fish','pred_fish_at_conf','TP_at_conf','FP_at_conf'])
        w.writerows(rows)
    print('[✓] Wrote CSV:', CSV_PATH)
```
