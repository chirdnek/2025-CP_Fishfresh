# 🐟 FishFresh — YOLO Fish Detector → TFLite (Clean Copy)

This notebook loads a **pretrained YOLOv8 (OpenImages‑V7) detector**, restricts outputs to the **Fish** class only, runs predictions on your **tray photos**, and exports **both** FP16 and FP32 **TFLite** models for your Flutter app.

**Pipeline:** Detector (Fish only) → crop ROIs → MobileNet classifier (species + freshness).

````python
# Versions (sanity check)
import sys
import ultralytics, onnxruntime, tensorflow as tf
print("Python:", sys.version)
print("Ultralytics:", ultralytics.__version__)
print("ONNXRuntime:", onnxruntime.__version__)
print("TensorFlow:", tf.__version__)
````

## 1) Folders & paths
Put a few tray images (jpg/png) in `assets/samples/` to test.


````python
from pathlib import Path

ROOT    = Path.cwd()
ASSETS  = ROOT/'assets';           ASSETS.mkdir(exist_ok=True)
SAMPLES = ASSETS/'samples';        SAMPLES.mkdir(parents=True, exist_ok=True)
OUT     = ROOT/'runs_fishdet';     OUT.mkdir(exist_ok=True)

print("ROOT   =", ROOT)
print("ASSETS =", ASSETS)
print("SAMPLES=", SAMPLES, "(put tray photos here)")
print("OUT    =", OUT)
````

## 2) Load YOLOv8 OpenImages‑V7 weights (auto-download) & resolve `fish_id`
- `yolov8n-oiv7.pt` → faster/smaller
- `yolov8s-oiv7.pt` → a bit more accurate (slower)


````python
from ultralytics import YOLO

MODEL_NAME = 'yolov8n-oiv7.pt'  # or 'yolov8s-oiv7.pt'
model = YOLO(MODEL_NAME)        # auto-downloads on first use
print('[✓] Loaded YOLO model:', MODEL_NAME)

# Resolve the class id for 'Fish' (case-insensitive)
fish_id = next((int(k) for k, v in model.names.items() if str(v).lower() == 'fish'), None)
assert fish_id is not None, "Fish class not found in this checkpoint."
print('Fish class id =', fish_id)
````

## 3) Predict on tray photos (Fish only)
- Tune `IMGSZ` (640–960), `CONF` (0.25–0.45), and `MAXDET` (allow many boxes).
- Results are saved to `runs_fishdet/pred/` (images + YOLO txt).


````python
IMGSZ, CONF, MAXDET = 640, 0.30, 300

from pathlib import Path
img_list = [*SAMPLES.glob('*.jpg'), *SAMPLES.glob('*.jpeg'), *SAMPLES.glob('*.png')]
print('Found', len(img_list), 'images in', SAMPLES)

if img_list:
    results = model.predict(
        source=str(SAMPLES),
        imgsz=IMGSZ,
        conf=CONF,
        max_det=MAXDET,
        classes=[fish_id],      # return Fish boxes only
        save=True,
        save_txt=True,
        project=str(OUT),
        name='pred',
        exist_ok=True,
        verbose=False,
        workers=0
    )
    print('[✓] Done. See annotated results in:', OUT/'pred')
else:
    print('[i] No images found in', SAMPLES)
````

## 4) Export **both** TFLite models (FP16 and FP32) + save a tiny config
- Files saved to `runs_fishdet/export/`:
  - `fishdet_fp16.tflite`
  - `fishdet_fp32.tflite`
  - `detector_config.json` (contains `imgsz`, `conf`, `fish_id`)


````python
from pathlib import Path
import shutil, json

EXPORT_DIR = OUT/'export'
EXPORT_DIR.mkdir(parents=True, exist_ok=True)

def _pick_tfl(p):
    seq = p if isinstance(p, (list, tuple)) else [p]
    for x in seq:
        if str(x).endswith('.tflite'):
            return Path(x)
    return None

print('[*] Exporting FP16 (half=True)...')
res16 = model.export(format='tflite', imgsz=IMGSZ, half=True, int8=False)
tfl16 = _pick_tfl(res16); assert tfl16 and tfl16.exists(), 'FP16 TFLite not found.'
fp16_out = EXPORT_DIR/'fishdet_fp16.tflite'
shutil.copy2(tfl16, fp16_out)
print('[✓] FP16 saved to:', fp16_out)

print('[*] Exporting FP32 (half=False)...')
res32 = model.export(format='tflite', imgsz=IMGSZ, half=False, int8=False)
tfl32 = _pick_tfl(res32); assert tfl32 and tfl32.exists(), 'FP32 TFLite not found.'
fp32_out = EXPORT_DIR/'fishdet_fp32.tflite'
shutil.copy2(tfl32, fp32_out)
print('[✓] FP32 saved to:', fp32_out)

# Save tiny config for the app
cfg = {'imgsz': IMGSZ, 'conf': float(CONF), 'fish_id': int(fish_id)}
(EXPORT_DIR/'detector_config.json').write_text(json.dumps(cfg, indent=2))
print('[✓] Wrote', EXPORT_DIR/'detector_config.json')

print('\nAll exports:')
print(' -', fp16_out)
print(' -', fp32_out)
````

## 5) (Optional) Quick TFLite load test
Just allocates the interpreter(s) and prints output tensor metadata.


````python
import tensorflow as tf

for p in [EXPORT_DIR/'fishdet_fp16.tflite', EXPORT_DIR/'fishdet_fp32.tflite']:
    print('\n[i] Testing:', p)
    interpreter = tf.lite.Interpreter(model_path=str(p))
    interpreter.allocate_tensors()
    outs = interpreter.get_output_details()
    print('[✓] Interpreter OK. Outputs:')
    for od in outs:
        print(' -', od.get('name'), od['shape'], od['dtype'])
````

## 6) (Optional) Quick evaluation on a labeled set (Precision/Recall/F1 @ IoU=0.50)

Place eval data here (YOLO format; class **0 = fish**):
```
datasets/eval/images/xxx.jpg
datasets/eval/labels/xxx.txt   # <class> <cx> <cy> <w> <h>  (normalized)
```
This gives a fast **P/R/F1** proxy to decide if the detector is acceptable. (True mAP needs a full PR curve.)


````python
from pathlib import Path
import numpy as np
from PIL import Image

EVAL_DIR = ROOT/'datasets'/'eval'
IMG_DIR  = EVAL_DIR/'images'
LBL_DIR  = EVAL_DIR/'labels'

if IMG_DIR.exists() and LBL_DIR.exists():
    def yolo_txt_to_boxes(lbl_path, img_w, img_h):
        boxes = []
        if not lbl_path.exists():
            return boxes
        for line in lbl_path.read_text().strip().splitlines():
            parts = line.strip().split()
            if len(parts) != 5: 
                continue
            cls, cx, cy, w, h = parts
            if int(float(cls)) != 0:  # class 0 only = fish
                continue
            cx, cy, w, h = map(float, (cx, cy, w, h))
            x1 = (cx - w/2) * img_w
            y1 = (cy - h/2) * img_h
            x2 = (cx + w/2) * img_w
            y2 = (cy + h/2) * img_h
            boxes.append([x1,y1,x2,y2])
        return boxes

    def iou(a, b):
        ax1, ay1, ax2, ay2 = a; bx1, by1, bx2, by2 = b
        inter_x1, inter_y1 = max(ax1, bx1), max(ay1, by1)
        inter_x2, inter_y2 = min(ax2, bx2), min(ay2, by2)
        iw, ih = max(0.0, inter_x2-inter_x1), max(0.0, inter_y2-inter_y1)
        inter = iw * ih
        ua = (ax2-ax1)*(ay2-ay1) + (bx2-bx1)*(by2-by1) - inter + 1e-6
        return inter / ua

    IOU_THR = 0.50
    img_paths = sorted([p for p in IMG_DIR.glob('*.jpg')] + [p for p in IMG_DIR.glob('*.jpeg')] + [p for p in IMG_DIR.glob('*.png')])
    if len(img_paths) == 0:
        print('[i] No images in', IMG_DIR)
    else:
        TP=FP=FN=0
        for p in img_paths:
            with Image.open(p) as im:
                W, H = im.size
            gt = yolo_txt_to_boxes(LBL_DIR/(p.stem+'.txt'), W, H)

            r = model.predict(source=str(p), imgsz=IMGSZ, conf=CONF, max_det=MAXDET, classes=[fish_id], verbose=False)[0]
            pred = (r.boxes.xyxy.cpu().numpy() if r.boxes is not None else np.empty((0,4), dtype=float)).tolist()

            used = set()
            for g in gt:
                best=-1; best_i=-1
                for i,b in enumerate(pred):
                    if i in used: 
                        continue
                    # compute IoU
                    ax1, ay1, ax2, ay2 = g
                    bx1, by1, bx2, by2 = b
                    inter_x1, inter_y1 = max(ax1, bx1), max(ay1, by1)
                    inter_x2, inter_y2 = min(ax2, bx2), min(ay2, by2)
                    iw, ih = max(0.0, inter_x2-inter_x1), max(0.0, inter_y2-inter_y1)
                    inter = iw * ih
                    ua = (ax2-ax1)*(ay2-ay1) + (bx2-bx1)*(by2-by1) - inter + 1e-6
                    iou_v = inter / ua
                    if iou_v > best:
                        best, best_i = iou_v, i
                if best >= IOU_THR and best_i >= 0:
                    TP += 1; used.add(best_i)
                else:
                    FN += 1
            FP += len(pred) - len(used)

        precision = TP / (TP + FP + 1e-6)
        recall    = TP / (TP + FN + 1e-6)
        f1        = 2*precision*recall / (precision + recall + 1e-6)

        print(f"Eval images: {len(img_paths)}")
        print(f"TP={TP}  FP={FP}  FN={FN}")
        print(f"Precision@0.50 = {precision*100:.2f}%")
        print(f"Recall@0.50    = {recall*100:.2f}%")
        print(f"F1@0.50        = {f1*100:.2f}%")

        print('MODEL STATUS:', '✅ OK to export' if f1 >= 0.80 else '❌ Consider switching/fine-tuning')
else:
    print('[i] Skipping eval: put labeled data under datasets/eval/{images,labels}')
````
